```python
import numpy as np
import wandas as wd

signal_a = wd.read('recording-a.wav')
signal_b = wd.read('recording-b.wav')
```

```python
# Replace only when a real calibration certificate/value is available.
# signal_a = signal_a.with_calibration({0: wd.ChannelCalibration(factor=0.42, unit='Pa')})
# signal_b = signal_b.with_calibration({0: wd.ChannelCalibration(factor=0.42, unit='Pa')})

for name, frame in {'A': signal_a, 'B': signal_b}.items():
    data = np.atleast_2d(np.asarray(frame.data))
    references = [channel.level_reference for channel in frame.channels]
    dc_offset = np.mean(data, axis=-1)
    clipping_screen = [
        {
            'channel': channel.label,
            'peak_fs': float(np.max(np.abs(data[index]))),
            'near_fs_fraction': float(np.mean(np.abs(data[index]) >= 0.999)),
            'possible_clipping': bool(np.max(np.abs(data[index])) >= 0.999),
        }
        if references[index].unit == 'dBFS' else None
        for index, channel in enumerate(frame.channels)
    ]
    print(name, frame.sampling_rate, frame.duration, frame.labels)
    print('references:', [reference.label for reference in references])
    print('rms:', frame.rms, 'crest:', frame.crest_factor, 'dc_offset:', dc_offset)
    print('full-scale clipping screen:', clipping_screen)
```
